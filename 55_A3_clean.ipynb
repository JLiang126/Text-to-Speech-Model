## 1.1.2 Audio preprocessing (document your choices)
- **Sampling rate:** `original`
  - Sample rate is how many times per seconds the microphone measures the air pressure, a sample rate of $s$ can only capture audio frequencies up to $\frac{s}{2}$. Natively the frequency is 22,050 Hz so it can capture up to 11,025 Hz which we have kept the same as this mimics the natural audible requency range for human speech.
- **Silence trimming / padding:** `-40 dB root mean squared energy threshold (20 ms window, 5 ms hop) and a 50 ms trailing zero pad`
  - The -40 dB root mean squared energy threshold is to cut out any noise
  - The trailing 50ms of 0 dB to give me the model a clear sense of the end of a sample.
- **Any filtering / normalization:** `DC Offset Removal & Peak Normalization`
  - DC Offset, waveform - waveform.mean(), cheap or slightly uncalibrated recording equipment can shift a soundwave slightly above or below zero so "silence" sits at 0.02 instead of 0.0, mean centering forces the wave back at zero so silent regions have zero energy.
  - Peak Normalization, waveform / max(abs(waveform)) * 0.95, every audio clip is scaled its loudest peak reaches 0.95. Without this, the neural network can confuse emphasis and volumes on certain words so we shrink the magnitude of volume by 95% to allow for headroom so convolutional filters don't clip (saturate) the wave at the boundaries.

## 1.1.3 Text preprocessing (minimum: justify your choice)
- **Normalisation:** `NFKC Unicode, Case Preservation and Number/Abbreviation Expansion`
  - LJSpeech's `metadata.csv` provides raw text in Column 2 and verbally expanded text in Column 3 where characters, abbreviations and numbers like `12` and `Mr.` are expanded into `twelve` and `Mister`. We pull exclusively from Column 3 as raw digits and symbols have no direct grapheme to phoneme mapping.
  - LJSpeech transcripts come from 19th century, non-fiction books, it contains curly quotes, em dashes, and accented characters, which are not part of a standardised unicode language, NFKC normalisation converts the unknown symbols into their standard equivalents. This allows for the network to onl train on a fixed, reduced, known set of characters rather than having a large set of obscure characters.
  - Parentheses such as`()` and `[]` appear in LJSpeech transcripts but are absent from the VITS vocabulary. When encountering parenthesis speakers should take a breath hence the `clean_normalised_text` preprocessor converts bracket boundaries into comma pauses ", " whilst preventing adjacent duplicate punctuation such as ",." or ",,".
  - Our pipeline passes text into the phonemiser before tokenisation, hence no lowercasing is done. Preserving uppercase letters and Unicode accents allows espeak to use capitalisation to identify acronyms, initials, and proper nouns before outputting purely lowercase IPA phonemes.
- **Tokeniser / phonemiser (if used):** `VitsTokenizer with eSpeak-ng Backend and <unk> ID 174 Collision Fix`
  - The pretrained checkpoint used in ylacombe/vits-vctk-with-discriminator uses Hugging Face's `VitsTokenizer` configured with the `espeak-ng` phonemiser. It converst English text into International Phonetic Alphabet (IPA) sequences with primary `ˈ` and secondary `ˌ` stress markers converts english into characters representing their . 
  - The original VITS 178-symbol table (`0–177`) had a duplicate apostrophe entry that left index `174` unused in `vocab.json`, the English vowel phoneme `ᵻ` occupies ID `177`, however an original bug in the code makes `VitsTokenizer` assign an unmapped `<unk>` token to `len(vocab) == 177`, aliasing `<unk>` onto `ᵻ` and stripping it during decoding. Our `load_vctk_tokenizer` helper binds `<unk>` to the duplicated slot at ID `174` preserving the ID `177` for `ᵻ` without resizing the pretrained `[178, 192]` embedding matrix.
  - The tokeniser intersperses the padding/blank token with `ID 0` representing a space `"_"` between every phoneme token and at both sequence boundaries so that the sequence has 2N + 1 length. This is required by VITS's Monotonic Alignment Search (MAS) to model phoneme transitions and brief silences.

# 1.2 End-to-end TTS pipeline

## 1.2.1 Pipeline overview

- **TTS approach:** We fine-tune an existing model instead of training from scratch. The starting point is `ylacombe/vits-vctk-with-discriminator`, a VITS model trained on the VCTK corpus (109 speakers) that comes with its discriminator, so adversarial training can continue on LJSpeech. MUST ADD markdown overview of the VITS + HiFi-GAN discriminator fine-tuning pipeline.
 
- **Key components:**  
  - Text front-end: the normalised LJSpeech text (1.1.3) is turned into IPA phonemes by espeak-ng and then into token IDs by the VCTK tokeniser, with a blank token between every symbol. 
  - Acoustic model: VITS (Kim, Kong & Son, 2021), a conditional VAE with a text encoder, a stochastic duration predictor, a normalising flow and a posterior encoder, aligned with monotonic alignment search. A learned speaker embedding (one 256-dimensional vector per VCTK speaker) conditions the model. Each run starts from one VCTK speaker and fine-tunes towards LJ. 
  - Vocoder waveform generator: the HiFi-GAN-style decoder inside VITS, trained end to end with the rest of the model, so there is no separate vocoder.  
- **Output:** a 22,050 Hz mono audio`.wav` file.

- **Python:** 3.14.8

## 1.2.3 Training commands

Every run uses the same training script and one file of settings:

`python -m scripts.train_model --config configs/trial_run.yaml --output-mode restart`

`configs/` holds the round-1 settings. The round-2 runs used the same script with the 80/10/10 split and a peak learning rate of 4e-5, and the full settings of each run are saved next to its logs in `outputs_v2/<experiment>/run/training_config.json`. Adding `--test-mode yes` runs one epoch on a small subset as a quick check.

When `scripts/train_model.py` executes, it coordinates a multi-stage adversarial training loop:
1. It loads `VitsModelForPreTraining` and `VitsDiscriminator` from `ylacombe/vits-vctk-with-discriminator`, setting `layerdrop = 0.0` and wrapping conv layers in `weight_norm`.
2. **Two-Player Step Optimization:**
   * **Discriminator Update:** Computes Least-Squares GAN (`disc_loss`) on real audio vs. generated audio segments and steps `disc_optimizer` with gradient clipping.
   * **Generator Update:** Computes the composite objective—80-band Log-Mel L1 loss (`mel`), Prior-Posterior KL divergence (`kl`), Duration Predictor negative log-likelihood (`duration`), Multi-Scale Feature Matching loss (`fmaps`), and Adversarial loss (`gen`).
3. **Artifact Logging:**
   * `outputs/test/run/training.log`: Chronological step-by-step console and VRAM logs.
   * `outputs/test/run/training_config.json`: Complete snapshot of hyperparameters and seed settings.
   * `outputs/test//run/*_loss_history.csv` & `loss_curves.png`: Exported metrics across epochs.
   * `outputs/test//intermediate_checkpoints/` & `<output_dir>/run/`: Exported inference-ready Hugging Face checkpoints with training weight normalizations stripped via `save_inference_checkpoint()`.

# 1.4 Evaluation, coding practice and attribution

## 1.4.1 Objective metrics

We score every system on the same 150 held-out sentences, the first 150 of the test split from 1.1. None of our runs saw these sentences in training or validation (checked in the next cell). The systems are the 11 round-2 runs from 1.3 and two reference points: the real LJ recordings (the best we can hope for) and the pretrained model before fine-tuning, with each starting speaker.

| Metric | What it measures | Better |
|---|---|---|
| MCD-DTW (dB) | spectral distance to the real recording of the same sentence | lower |
| WER (%) | word errors in a Whisper small.en transcript, pooled over all sentences | lower |
| Speaker similarity | ECAPA cosine similarity to the centroid (mean embedding) of 50 LJ training clips. It reflects voice identity (timbre, pitch range, speaking style), not the words or their timing | higher |
| UTMOS | predicted naturalness on a 1 to 5 scale | higher |
| Pitch (Hz) | median F0 of each sentence (LJ's recordings average 207 Hz) | closer to LJ |
| Duration ratio | synthetic length over real length of the same sentence | closer to 1 |

We also time synthesis on the 12 prompts (latency and real-time factor). To decide whether a change made a difference, we compare two runs sentence by sentence. A difference counts only if its 95% bootstrap confidence interval excludes zero and a Wilcoxon signed-rank test, Holm-corrected for the number of comparisons, gives p < 0.05.

The code is in `scripts/`: `evaluate_experiments.py` runs the whole evaluation (`evaluate.py` scores one system, `compare.py` runs the paired tests and `scripts/evaluation/` holds the metrics), and `report_summary.py` makes the tables and figures below.

**What the results show**

- **Starting voice:** starting from speaker 24, the VCTK voice furthest from LJ (W), beat starting from speaker 7, the closest (B). W is clearly better on MCD, speaker similarity, UTMOS (+0.72) and pitch, and worse on nothing, which is why it is our submitted system.
- **Embedding only:** training only the speaker embedding leaves the voice where it started. Speaker similarity stays at 0.20 for E and −0.11 for EW (LJ's own recordings score 0.85), and E's pitch only moves from 117 to 127 Hz. The voice changes when the rest of the network is fine-tuned.
- **10 times the embedding learning rate:** this moves the embedding-only runs further towards LJ (E10 reaches a similarity of 0.30 and a pitch of 164 Hz) but makes no clear difference to full fine-tuning. W10 matches W on every metric, and B10 gains 0.01 dB of MCD while its pitch drifts 6 Hz further from LJ.
- **Text encoder:** freezing it (F) is clearly worse than B on MCD, speaker similarity and duration, while the layer drop of the original VITS recipe (O) gives only small, mixed changes.

## 1.4.3 Coding practice and clarity

The code is split by job:

| Folder | What is in it |
|---|---|
| `scripts/` | data (`DataConfig`, `DataLoader`, `AudioProcessor`, `TextProcessor`), training (`train_model`, `model`, `configuration_vits`, `monotonic_alignment`), inference (`synth`) and evaluation (`evaluate`, `compare`, `evaluate_experiments`, `report_summary`, `pretrained_validation_losses`, `rank_speakers`, `data_checks`) |
| `scripts/evaluation/` | the metrics, statistics, checkpoint checks and model loading, explained in its README |
| `configs/`, `experiments/` | the round-1 training settings, and the run tables and report plans the evaluation reads |
| `outputs/`, `outputs_v2/` | the training logs and the settings each run used |
| `results/` | every score, report and figure |
| `tests/` | unit tests that run offline: `python -m pytest tests -q` |

For reproducibility, one seed (42) fixes the split, the training randomness and the batch order. Each training run saves the settings it actually used, and the evaluation checks every checkpoint against them and against the pretrained weights before scoring it (for example, that a frozen module really did not change). Scoring uses a fixed sampling seed per sentence, treats real and synthetic audio the same way, refuses to score a model on sentences from its own training split and can resume after a stop. This notebook reloads the saved scores instead of repeating hours of scoring.
